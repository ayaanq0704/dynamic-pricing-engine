```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')

# Load the dataset
df = pd.read_csv("C:/Users/Ayaan/Desktop/dynamic_pricing_project/data/raw/hotel_bookings.csv")

print(f"Dataset shape: {df.shape}")
print(f"Columns: {list(df.columns)}")

# Basic data exploration
print("\n=== BASIC INFO ===")
print(df.info())
print("\n=== PRICE STATISTICS ===")
print(df['adr'].describe())

# Key insights for pricing
print("\n=== PRICING INSIGHTS ===")
print(f"Average Daily Rate (ADR): ${df['adr'].mean():.2f}")
print(f"Price range: ${df['adr'].min():.2f} - ${df['adr'].max():.2f}")
print(f"Cancellation rate: {df['is_canceled'].mean()*100:.1f}%")

# Visualize price distributions
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Price distribution
axes[0,0].hist(df['adr'], bins=50, alpha=0.7)
axes[0,0].set_title('Price Distribution (ADR)')
axes[0,0].set_xlabel('Average Daily Rate ($)')

# Price by hotel type
df.boxplot(column='adr', by='hotel', ax=axes[0,1])
axes[0,1].set_title('Price by Hotel Type')

# Price by month (seasonality)
monthly_prices = df.groupby('arrival_date_month')['adr'].mean()
axes[1,0].plot(monthly_prices.index, monthly_prices.values, marker='o')
axes[1,0].set_title('Average Price by Month')
axes[1,0].set_xlabel('Month')
axes[1,0].set_ylabel('Average Daily Rate ($)')

# Lead time vs price
axes[1,1].scatter(df['lead_time'], df['adr'], alpha=0.5)
axes[1,1].set_title('Lead Time vs Price')
axes[1,1].set_xlabel('Lead Time (days)')
axes[1,1].set_ylabel('Average Daily Rate ($)')

plt.tight_layout()
plt.show()

# Demand patterns
print("\n=== DEMAND PATTERNS ===")
demand_by_month = df.groupby('arrival_date_month').size()
print("Bookings by month:")
print(demand_by_month)

# Price elasticity analysis (simplified)
print("\n=== PRICE ELASTICITY INSIGHTS ===")
price_buckets = pd.cut(df['adr'], bins=10, labels=False)
demand_by_price = df.groupby(price_buckets).size()
print("Bookings by price bucket (0=lowest, 9=highest):")
print(demand_by_price)
```
